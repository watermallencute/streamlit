# **1. Context**

Industri perhotelan menghadapi tantangan struktural pada sisi **revenue management** karena kamar hotel merupakan *perishable inventory*, yaitu kapasitas yang nilainya terikat pada malam tertentu dan tidak dapat disimpan untuk dijual pada periode berikutnya. Ketika satu malam telah berlalu dan kamar tidak terjual, kesempatan untuk menghasilkan pendapatan dari kamar tersebut hilang secara permanen. Berbeda dengan produk fisik yang dapat disimpan sebagai persediaan dan dijual di kemudian hari, kapasitas kamar hotel memiliki nilai yang hanya dapat direalisasikan pada periode waktu tertentu.

Kondisi ini membuat kamar yang tidak terisi menimbulkan dampak finansial dari beberapa sisi. **Pertama**, terdapat pendapatan yang gagal direalisasikan. Setelah tanggal menginap terlewati, potensi pendapatan dari kamar tersebut tidak dapat dipulihkan atau dialihkan ke periode berikutnya. **Kedua**, hotel tetap menanggung biaya operasional meskipun kamar tidak terisi. Biaya seperti gaji staf, housekeeping, listrik, dan pemeliharaan fasilitas sebagian besar bersifat tetap (*fixed cost*) atau tidak sepenuhnya bergantung pada tingkat okupansi. Akibatnya, kamar kosong menciptakan kondisi ketika biaya tetap dikeluarkan tanpa adanya pendapatan yang dihasilkan dari kamar tersebut.

**Ketiga**, dampak kehilangan pendapatan dapat menjadi lebih besar pada periode dengan permintaan tinggi (*peak season*). Ketika kapasitas hotel telah dialokasikan berdasarkan reservasi yang ada, permintaan dari calon tamu lain dapat ditolak karena kamar dianggap telah penuh. Apabila reservasi tersebut kemudian dibatalkan, hotel tidak hanya kehilangan pendapatan dari kamar yang kosong, tetapi juga kehilangan kesempatan untuk menjual kamar tersebut kepada tamu lain yang sebelumnya tidak dapat melakukan reservasi. Kondisi ini menciptakan *opportunity cost* yang berpotensi lebih besar dibandingkan kehilangan pendapatan pada periode dengan tingkat permintaan rendah.

Pada data historis yang digunakan dalam proyek ini, **cancellation rate gabungan tercatat sekitar 37%**, dengan cancellation rate pada Resort Hotel sebesar **27.8%**, sedangkan cancellation rate pada City Hotel sebesar **41.7%**. Tingkat pembatalan yang cukup tinggi ini menunjukkan bahwa cancellation merupakan risiko yang signifikan dalam proses reservasi hotel. Perbedaan cancellation rate antarhotel juga menunjukkan bahwa tingkat risiko pembatalan dapat bervariasi berdasarkan karakteristik hotel. Kondisi tersebut menunjukkan adanya kebutuhan untuk mengidentifikasi reservasi yang memiliki risiko cancellation lebih tinggi sejak reservasi dibuat, terutama karena setiap pembatalan dapat berdampak pada tingkat okupansi dan potensi pendapatan hotel.

Proyek ini berfokus pada **prediksi kemungkinan pembatalan reservasi (`is_canceled`)** menggunakan data historis pemesanan hotel di Portugal, yang mencakup **Resort Hotel** di Algarve dan **City Hotel** di Lisbon. Tujuan utamanya adalah memberikan estimasi risiko cancellation pada tingkat reservasi sehingga hotel dapat mengambil tindakan preventif berdasarkan tingkat risiko masing-masing reservasi, bukan memperlakukan seluruh reservasi dengan pendekatan yang sama.

**Stakeholder utama proyek ini:**

- **Reservation Team** — membutuhkan identifikasi reservasi dengan risiko cancellation yang lebih tinggi untuk menentukan tindakan preventif secara terarah, seperti *reconfirmation*, sebelum tanggal kedatangan.

Saat ini, hotel belum memiliki mekanisme sistematis untuk membedakan reservasi dengan risiko cancellation tinggi dan rendah. Pendekatan yang berjalan cenderung bersifat **reaktif**, di mana tindakan preventif, apabila dilakukan, baru dilakukan mendekati tanggal check-in. Padahal, semakin dekat dengan tanggal kedatangan, semakin terbatas pula waktu yang tersedia bagi hotel untuk mengantisipasi pembatalan dan menjual kembali kamar yang kosong.

Proyek ini bertujuan mengubah pendekatan tersebut menjadi **proaktif** dengan memanfaatkan atribut yang tersedia sejak reservasi dibuat untuk memprediksi risiko cancellation lebih awal. Dengan demikian, model dapat memberikan sinyal risiko yang dapat digunakan oleh tim bisnis untuk menentukan prioritas intervensi dan mengambil tindakan sebelum tanggal kedatangan, sehingga peluang kehilangan pendapatan akibat cancellation dapat diminimalkan.

# **2. Problem Statements**

**Permasalahan Utama**

Saat ini, hotel tidak memiliki cara sistematis untuk mengetahui reservasi mana yang berisiko tinggi dibatalkan. Kondisi ini dapat menyebabkan:

- **Revenue loss**, karena kamar yang dibatalkan berpotensi tidak sempat dijual kembali.
- **Occupancy forecasting yang kurang akurat**, karena jumlah booking tidak selalu mencerminkan jumlah tamu yang benar-benar datang.
- **Perencanaan operasional yang kurang optimal**, seperti kebutuhan housekeeping, F&B, inventory, dan front office.
- **Strategi overbooking yang tidak terukur**, karena hotel sulit memperkirakan berapa banyak booking yang kemungkinan tidak terealisasi.
- **Intervensi yang tidak terarah**, karena seluruh reservasi cenderung diperlakukan dengan pendekatan yang sama meskipun memiliki tingkat risiko cancellation yang berbeda.


**Rumusan Masalah (Business Question)**

Analisis ini bertujuan menjawab beberapa pertanyaan bisnis utama:

1. `Profile:`  Karakteristik reservasi seperti apa yang memiliki kecenderungan cancellation paling tinggi?

2. `Drivers:`  Faktor apa saja yang paling berpengaruh terhadap kemungkinan sebuah reservasi dibatalkan?

3. `Strategy:` Bagaimana hotel dapat merancang strategi preventif yang tepat untuk mengurangi potensi revenue loss dan meningkatkan kepastian okupansi?

4. `Targeting:`Reservasi dengan karakteristik dan probabilitas cancellation seperti apa yang harus diprioritaskan untuk mendapatkan tindakan preventif dari hotel?

5. `Impact:`   Bagaimana hasil prediksi cancellation dapat digunakan untuk meningkatkan occupancy forecasting, mengoptimalkan inventory kamar, dan mendukung strategi overbooking yang lebih terukur?



# **3. Goals**

Tujuan utama proyek ini adalah membangun model machine learning klasifikasi yang mampu memprediksi probabilitas sebuah reservasi akan dibatalkan, berdasarkan atribut yang tersedia pada saat reservasi dibuat (bukan atribut yang baru diketahui setelah tamu check-in/check-out).

Dengan model ini, hotel diharapkan dapat:
* **Mengidentifikasi reservasi beresiko tinggi dibatalkan** sejak awal, sehingga tim reservasi dapat melakukan tindakan preventif yang lebih terarah berdasarkan tingkat resiko setiap reservasi seperti konfirmasi ulang atau menawarkan kebijakan deposit yang sesuai.
* **Meningkatkan akurasi forecast okupansi**, sehingga perencanaan housekeeping, staf, dan strategi harga menjadi lebih presisi.
* **Mendukung strategi overbooking yang lebih terukur** berdasarkan probabilitas pembatalan per segmen, bukan asumsi rata-rata historis semata.
* **Menurunkan potensi revenue loss** akibat kamar yang batal terpakai namun tidak sempat dijual kembali.

## Tahap 6 Model Evaluation & Threshold Tunning

Target hotel booking cancellation memiliki moderate class imbalance. Oleh karena itu, **ROC-AUC** digunakan sebagai metrik utama untuk pengembangan dan perbandingan model. **ROC-AUC** mengukur kemampuan model dalam membedakan reservasi yang akan dibatalkan dan yang tidak dibatalkan pada berbagai kemungkinan classification threshold. Semakin tinggi nilai **ROC-AUC**, semakin baik kemampuan model dalam memberikan skor risiko yang lebih tinggi kepada reservasi yang benar-benar dibatalkan.

Setelah model terbaik dipilih berdasarkan **ROC-AUC**, dilakukan threshold tuning menggunakan **Accuracy**. **Accuracy** mengukur proporsi keseluruhan prediksi yang benar, baik pada reservasi yang benar-benar dibatalkan maupun yang benar-benar terealisasi, terhadap seluruh reservasi yang dievaluasi. Threshold dengan **Accuracy** tertinggi dipilih agar hotel memperoleh titik keseimbangan yang wajar antara risiko `revenue loss (False Negative)` dan `risiko walk situation/biaya intervensi yang berlebihan (False Positive)`, sehingga tindakan preventif dapat dilakukan secara lebih tepat sasaran dan efisien.

# **5. Metric Evaluation**

Agar model machine learning benar-benar memberikan nilai bisnis, evaluasi teknis (ML metrics) perlu diterjemahkan ke dalam konteks biaya operasional hotel (business cost).

## **5.1 Business Cost of Misclassification**
* **False Positive (Tipe I Error):** Model memprediksi reservasi akan `dibatalkan`, padahal tamu tersebut `benar-benar datang`.
  * **Business Cost:** Jika hotel bertindak agresif (misalnya melakukan overbooking berdasarkan prediksi ini), risiko yang muncul adalah *walk situation* — tamu yang datang tidak mendapat kamar, sehingga hotel harus membayar kompensasi (relokasi ke hotel lain, biaya transportasi) dan berisiko kehilangan reputasi/loyalitas tamu.
* **False Negative (Tipe II Error):** Model memprediksi reservasi `tidak akan dibatalkan`, padahal tamu tersebut `benar-benar membatalkan`.
  * **Business Cost:** Hotel kehilangan kesempatan untuk melakukan tindakan pencegahan (follow-up, penawaran deposit) atau menjual ulang kamar tersebut ke tamu lain, sehingga kamar berpotensi kosong dan pendapatan hilang begitu saja.

## **5.2 Machine Learning Metrics & Business Alignment**
* **ROC-AUC (Primary Optimization Metric)**
  * **Business Meaning:** Mengukur kemampuan model dalam membedakan reservasi yang akan `cancel` dan `tidak cancel` di seluruh kemungkinan *threshold*, dengan mempertimbangkan performa pada kedua kelas secara seimbang `(True Positive Rate vs. False Positive Rate)`. Metrik ini dipilih sebagai metrik utama karena memberikan gambaran menyeluruh tentang kemampuan diskriminatif model, sekaligus tetap relevan digunakan pada kondisi class imbalance yang bersifat **moderate (± 62,9% Not Canceled : 37,1% Canceled)** seperti pada dataset ini.
  * **Peran dalam Proyek:** Digunakan sebagai metrik utama pada saat *cross-validation* dan model comparison untuk memilih arsitektur model terbaik.
* **Accuracy (Threshold Tuning)**
  * **Business Meaning:** Mengukur proporsi keseluruhan prediksi yang benar, baik reservasi yang benar-benar `cancel` maupun yang benar-benar `tidak cancel` terhadap seluruh reservasi yang dievaluasi, sehingga merepresentasikan performa model secara umum pada kedua kelas sekaligus.
  * **Peran dalam Proyek:** Digunakan sebagai dasar penentuan optimal threshold klasifikasi, sehingga hotel memperoleh titik keseimbangan yang wajar antara risiko `revenue loss (False Negative)` dan `risiko walk situation/biaya intervensi yang berlebihan (False Positive)`, tanpa condong berlebihan ke salah satu jenis kesalahan.

# **6. Success Criteria**

## 1. Machine Learning Performance Benchmarks

Proyek ini dinyatakan berhasil apabila model prediktif memenuhi ambang batas berikut:

* **ROC-AUC ≥ 0,80 dan lebih tinggi dari baseline** model harus menghasilkan nilai yang lebih tinggi dibandingkan best baseline model, menunjukkan peningkatan kemampuan model dalam membedakan reservasi cancel dan tidak cancel di seluruh threshold.
* **Accuracy ≥ 75%** dioptimalkan melalui threshold tuning untuk memperoleh proporsi prediksi benar tertinggi secara keseluruhan, sehingga biaya bisnis akibat **False Positive** maupun **False Negative** dapat ditekan secara proporsional tanpa condong berlebihan ke salah satu jenis kesalahan.
* **Generalization Check (No Significant Overfitting):** Perbedaan performa antara training set dan testing set tidak boleh menunjukkan gap yang signifikan. Sehingga model dapat mempertahankan performa yang stabil pada data reservasi yang belum pernah dilihat sebelumnya.
* Model harus menggunakan **fitur yang tersedia pada saat reservasi dibuat**, bebas dari *data leakage* (misalnya status reservasi final atau tanggal check-out), agar dapat diterapkan secara real-time di masa mendatang.

## 2. Business & Operational Benchmarks

Performa teknis model harus diterjemahkan menjadi manfaat nyata bagi operasional dan pengambilan keputusan hotel:

- **Cancellation Detection Improvement:**  
  Penggunaan model harus mampu mengidentifikasi reservasi beresiko cancellation secara lebih efektif dibandingkan pendekatan baseline tanpa risk prediction.

- **Occupancy Forecasting Improvement:**  
  Cancellation probability yang dihasilkan model harus dapat memberikan informasi tambahan untuk memperkirakan jumlah booking yang kemungkinan benar-benar terealisasi, sehingga forecast okupansi menjadi lebih akurat.

- **Revenue Protection:**  
  Simulasi bisnis harus menunjukkan bahwa penggunaan cancellation risk dapat membantu hotel mengidentifikasi potential room inventory yang berisiko tidak terealisasi lebih awal, sehingga memberikan kesempatan untuk melakukan tindakan preventif atau menjual kembali kamar yang berpotensi kosong.

- **Operational Efficiency:**  
  Model harus memungkinkan hotel memprioritaskan tindakan seperti reconfirmation, reminder, atau monitoring hanya kepada reservasi yang memiliki risiko lebih tinggi.


- **Model Explainability & Transparency:**  
  Model final harus dapat diidentifikasi faktor-faktor utama yang berkontribusi terhadap risiko cancellation sehingga hasil model dapat dipahami dan digunakan oleh tim Revenue Management, Reservation, dan Operations.

Jika seluruh kriteria tersebut terpenuhi, model dapat dipertimbangkan untuk digunakan sebagai **Decision Support Tool** dalam proses **cancellation risk management, occupancy planning, dan room inventory management** hotel.
